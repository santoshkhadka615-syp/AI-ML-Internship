```python
import numpy as np
import pandas as pd
import seaborn as sns

pd.set_option('display.max_columns', None)
```

2. Load the Raw Dataset
We load the Titanic dataset from seaborn's built-in dataset loader. This is our raw data before any cleaning.

```python
raw_df = sns.load_dataset('titanic')
print('Shape of raw data:', raw_df.shape)
raw_df.head()
```

3. Explore the Raw Data
Check data types, missing values, and basic structure before cleaning.

```python
raw_df.info()
```

```python
raw_df.isnull().sum().sort_values(ascending=False)
```

```python
raw_df.describe()
```

4. Cleaning Step 1 - Drop Columns With Too Many Missing Values
The deck column is missing for most rows, so we drop it.

```python
clean_df = raw_df.copy()
clean_df = clean_df.drop(columns=['deck'])
print('Columns after dropping deck:', clean_df.columns.tolist())
```

5. Cleaning Step 2 - Handle Missing Values
Fill missing age with the median, and missing embarked/embark_town with the mode.

```python
median_age = clean_df['age'].median()
clean_df['age'] = clean_df['age'].fillna(median_age)

clean_df['embarked'] = clean_df['embarked'].fillna(clean_df['embarked'].mode()[0])
clean_df['embark_town'] = clean_df['embark_town'].fillna(clean_df['embark_town'].mode()[0])

clean_df[['age', 'embarked', 'embark_town']].isnull().sum()
```

6. Filtering Rows
Filter to survivors, and separately, 1st class passengers.



```python
survivors = clean_df[clean_df['survived'] == 1]
print('Number of survivors:', len(survivors))

first_class = clean_df[clean_df['class'] == 'First']
print('Number of 1st class passengers:', len(first_class))
```

7. GroupBy - Aggregating Data
Average age and survival rate grouped by passenger class.

```python
grouped = clean_df.groupby('class').agg(
    avg_age=('age', 'mean'),
    survival_rate=('survived', 'mean'),
    passenger_count=('survived', 'count')
)
grouped
```

8. Merging Data
Create a small lookup table mapping embark town to region, then merge it in.

```python
region_lookup = pd.DataFrame({
    'embark_town': ['Southampton', 'Cherbourg', 'Queenstown'],
    'region': ['England', 'France', 'Ireland']
})

clean_df = clean_df.merge(region_lookup, on='embark_town', how='left')
clean_df[['embark_town', 'region']].drop_duplicates()
```

9. Pivot Table
Survival rate by passenger class and sex.



```python
pivot = pd.pivot_table(
    clean_df,
    values='survived',
    index='class',
    columns='sex',
    aggfunc='mean'
)
pivot
```

10. Final Clean Dataset
Confirm the dataset is analysis-ready.

```python
print('Final shape:', clean_df.shape)
print(clean_df.isnull().sum().sort_values(ascending=False).head())
clean_df.head()
```

11. Save the Clean Dataset

```python
clean_df.to_csv('titanic_clean.csv', index=False)
print('Saved cleaned dataset to titanic_clean.csv')
```
